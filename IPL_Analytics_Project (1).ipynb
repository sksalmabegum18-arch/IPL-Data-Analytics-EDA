```python
# =========================
# Import Required Libraries
# =========================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

import warnings
warnings.filterwarnings('ignore')

# Display all columns
pd.set_option('display.max_columns', None)

print("Libraries Imported Successfully")
```

```python
# =========================
# Load Datasets
# =========================

matches = pd.read_csv(r"C:\Users\kasee\Downloads\matches.csv")
deliveries = pd.read_csv(r"C:\Users\kasee\Downloads\deliveries.csv")

print("Datasets Loaded Successfully")
```

```python
# =========================
# Handling Missing Values
# =========================

matches['winner'].fillna('No Result', inplace=True)

matches['player_of_match'].fillna('Not Awarded', inplace=True)

matches['city'].fillna('Unknown', inplace=True)
```

```python
matches.isnull().sum()
```

```python
# =========================
# Standardize Team Names
# =========================

matches.replace('Delhi Daredevils', 'Delhi Capitals', inplace=True)
matches.replace('Deccan Chargers', 'Sunrisers Hyderabad', inplace=True)

deliveries.replace('Delhi Daredevils', 'Delhi Capitals', inplace=True)
deliveries.replace('Deccan Chargers', 'Sunrisers Hyderabad', inplace=True)
```

```python
# =========================
# Unique Teams
# =========================

matches['team1'].unique()
```

```python
# =========================
# Save Cleaned Data
# =========================

matches.to_csv(r"C:\Users\kasee\Downloads\cleaned_matches.csv", index=False)

deliveries.to_csv(r"C:\Users\kasee\Downloads\cleaned_deliveries.csv", index=False)

print("Cleaned datasets saved successfully")
```

```python
# =========================
# Top Run Scorers
# =========================

top_batsmen = deliveries.groupby('batter')['batsman_runs'].sum().sort_values(ascending=False)

top_batsmen.head(10)
```

```python
# =========================
# Top Batsmen Visualization
# =========================

plt.figure(figsize=(12,6))

top_batsmen.head(10).plot(kind='bar')

plt.title("Top 10 Run Scorers in IPL", fontsize=16)
plt.xlabel("Players")
plt.ylabel("Runs Scored")

plt.xticks(rotation=45)

plt.show()
```

```python
# =========================
# Most Sixes
# =========================

most_sixes = deliveries[deliveries['batsman_runs'] == 6]

most_sixes['batter'].value_counts().head(10)
```

```python
# =========================
# Most Fours
# =========================

most_fours = deliveries[deliveries['batsman_runs'] == 4]

most_fours['batter'].value_counts().head(10)
```

```python
# =========================
# Most Fours Visualization
# =========================

plt.figure(figsize=(12,6))

most_fours['batter'].value_counts().head(10).plot(kind='bar')

plt.title("Players with Most Fours", fontsize=16)
plt.xlabel("Players")
plt.ylabel("Number of Fours")

plt.xticks(rotation=45)

plt.show()
```

```python
# =========================
# Top Wicket Takers
# =========================

top_bowlers = deliveries[deliveries['is_wicket'] == 1]

top_bowlers = top_bowlers['bowler'].value_counts()

top_bowlers.head(10)
```

```python
# =========================
# Toss Winner vs Match Winner
# =========================

toss_match_win = (matches['toss_winner'] == matches['winner']).mean() * 100

print(f"Toss Winner won the match {toss_match_win:.2f}% times")
```

```python
# =========================
# Merge Season Column
# =========================

deliveries = deliveries.merge(
    matches[['id', 'season']],
    left_on='match_id',
    right_on='id',
    how='left'
)
```

```python
# =========================
# Total Runs Per Season
# =========================

season_runs = deliveries.groupby('season')['total_runs'].sum()

season_runs
```

```python
# =========================
# Runs Per Season Visualization
# =========================

plt.figure(figsize=(12,6))

season_runs.plot(kind='line', marker='o')

plt.title("Total Runs Scored Per IPL Season", fontsize=16)
plt.xlabel("Season")
plt.ylabel("Runs")

plt.grid(True)

plt.show()
```

```python
# =========================
# Match Win Distribution
# =========================

plt.figure(figsize=(12,6))

sns.countplot(y='winner', data=matches,
              order=matches['winner'].value_counts().index)

plt.title("Match Wins Distribution", fontsize=16)

plt.show()
```

```python
# =========================
# Correlation Heatmap
# =========================

numeric_data = deliveries.select_dtypes(include=np.number)

plt.figure(figsize=(12,8))

sns.heatmap(numeric_data.corr(),
            annot=True,
            cmap='coolwarm')

plt.title("Correlation Heatmap", fontsize=16)

plt.show()
```

```python
# =========================
# Highest Team Scores
# =========================

team_scores = deliveries.groupby(
    ['match_id', 'batting_team']
)['total_runs'].sum().reset_index()

highest_scores = team_scores.sort_values(
    by='total_runs',
    ascending=False
)

highest_scores.head(10)
```

```python
# =========================
# Highest Team Scores Plot
# =========================

plt.figure(figsize=(12,6))

sns.barplot(
    x='batting_team',
    y='total_runs',
    data=highest_scores.head(10)
)

plt.title("Highest Team Scores in IPL", fontsize=16)
plt.xlabel("Teams")
plt.ylabel("Runs")

plt.xticks(rotation=45)

plt.show()
```
